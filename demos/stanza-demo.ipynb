```python
import numpy as np
import stanza
stanza.download('sk')
```

```python

nlp = stanza.Pipeline('sk', processors='tokenize,mwt,pos,lemma')
```

```python
text1 = "Rúry sú jednosmernou komunikačnou metódou. Dáta zapísané na jednom konci potrubia sa čítajú na druhom konci potrubia (jednosmerný tok údajov). Potrubia sa vytvárajú volanim jadra pipeline(). Keď je vytvorený kanál, systém zaberá 2 miesta v otvorenej tabuľke súborov procesu. Každý potomok zdedi takto vytvorenú fajku."
text2 = ("Rúra je jednosmerný komunikačný prostriedok. Údaje zapisované na jednom konci rúry sú prečítané na druhom konci rúry (jednosmerný tok dát). Rúra je vytvorená vola-nim jadra pipe(). Pri vytvorení rúry systém obsadí 2 pozície tabuľky otvorených súborov procesu. Takto vzniknutú rúru dedi každý potomok")

```

```python
doc1 = nlp(text1)
doc2 = nlp(text2)
```

```python
dir(doc1)
```

```python
print(doc1.entities)
```

```python

print(*[f'word: {word.text}\tupos: {word.upos}\txpos: {word.xpos}\tfeats: {word.feats if word.feats else "_"}' for sent in doc2.sentences for word in sent.words], sep='\n')
```
